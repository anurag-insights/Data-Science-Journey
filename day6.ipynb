#### Series
A Pandas 1D labeled array that can hold any data type, Think of it like a single column in spreadsheet (1-Dimensional)

```python
data = ['A', 'N', 'U', 'R', 'A', 'G']
series = pd.Series(data)
print(series)
```

```python
data = [True, False, True, False]
series = pd.Series(data)
print(series)
```

```python
data = [102, 104, 106]
series = pd.Series(data, index = ['a', 'b', 'c'])
print(series)

# index = ['a', 'b', 'c'] vaneko index (0,1,2) ko satta (a,b,c) rakheko
# NOTE: index ko length ra data ko length same hunu parxa natra error aauxa.
```

```python
print(series.loc['c'])
#loc = 'Location by Label'
# loc used to 'access the lock property'
#series.loc['c'] = 'c' index/location ma vako data access gareko ho
```

```python
series.loc['c'] = 200
print(series)

#'c' index ko value change vayera 200 vayo
```

```python
data = [100, 102, 104, 200, 202]
series = pd.Series(data, index = ['a', 'b', 'c', 'd', 'e'])
print(series >= 200) #returns boolean value (True/False)
```

```python
print(series[series>=200]) # returns the exact value which are >=200
```

 Using dictionaries instead of python list

```python
calories = {'Day 1': 1000, 'Day 2': 1200, 'Day 3': 1570}
series = pd.Series(calories)
print(series)
```

```python
series.loc['Day 3'] += 500
print(series.loc['Day 3'])
```

```python
print(series)
```

```python
print(series[series >= 2000])
```

 #### DataFrames
 A tabular data structure with rows and columns. (2-Dimensional), Similar to an excel spreadsheet.

```python
data = {'Name': ['Anurag', 'Ram', 'Shyam', 'Sita', 'Hari', 'Radha'],
         'Age': [20, 23, 24, 21, 35, 26]
       }
df = pd.DataFrame(data)
print(df)
```

```python
df = pd.DataFrame(data, index = ['Employee 1', 'Employee 2', 'Employee 3', 'Employee 4', 'Employee 5', 'Employee 6'])
print(df)
```

```python
print(df.loc['Employee 1'])
```

```python
 #Adding a new column
df['Job'] = ['Student', 'AI Engineer', 'N/A', 'Web Dev', 'ML Engineer', 'Student']

print(df)
```

```python
 #Adding a new row
new_row = pd.DataFrame([{'Name': 'Shiva', 'Age': 27, 'Job': 'Data Engineer'}], 
                        index = ['Employee 7']
                      )
df = pd.concat([df, new_row])
print(df)
```

#### Filtering
keeping the rows that match a condition

```python
tall_pokemon = df[df['Height']>=2]
print(tall_pokemon)
#returns the details of pokemons whose height are more than 2.
```

```python
heavy_pokemon = df[df['Weight']> 100]
print(heavy_pokemon)
```

```python
legend_pokemon = df[df['Legendary'] == 1]
print(legend_pokemon)
```

```python
water_pokemon = df[(df['Type1'] == 'Water') &
                    (df['Type2'] == 'Water')
    
    ]
print(water_pokemon)
```

#### Aggregation
Reduces a set of values into a single summary value used to summarize and analyze data,
often used with the groupby() function

```python
print(df)
```

```python
print(df.mean(numeric_only = True))
# returns the values which are only integers like no, height, weight.
```

```python
print(df.sum(numeric_only = True))
```

```python
print(df.min(numeric_only = True))
```

```python
print(df.max(numeric_only = True))
```

```python
print(df.count())
```

 ##### Selecting a single column instead of whole dataframe

```python
print(df['Height'].mean())

#Since 'Height is a numeric column, we don't need to write 'numeric_only = True'
```

```python
print(df['Legendary'].max())
```

##### Grouping the dataframes

```python
group = df.groupby('Type1')
print(group['Height']. mean())
#grouping the mean height of each types of pokemon.
```

```python
group = df.groupby('Type1')
print(group['Height'].sum())
#grouping the total height of each types of pokemon.
```
