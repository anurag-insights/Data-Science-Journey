#### Numeric data types
Python has three main built-in numeric types.
##### A. Integer (int)
An integer is a whole number without a decimal component.

```python
age = 20
students = 45
temperature = -5
score = 0

print(type(age))
print(type(temperature))
```

##### We can perform mathematical operations on integers

```python
a = 10
b = 3

print(a + b)   # Addition
print(a - b)   # Subtraction
print(a * b)   # Multiplication
print(a / b)   # Division
print(a // b)  # Floor division
print(a % b)   # Remainder
print(a ** b)  # Power
```

 ##### NOTE: / produces a floating-point result, even when the division is exact. // performs floor division.

Python integers can grow beyond the usual 32-bit or 64-bit integer ranges, subject to available memory.

#### B. Floating-point (float)
A float represents a number with a fractional or decimal component.

```python
price = 199.99
gpa = 3.45
height = 1.75
discount = 0.0

print(type(price))
print(type(gpa))
```

##### Floats support mathematical operations just like integers.

```python
price = 250.0
discount = 25.0

final_price = price - discount

print(final_price)  
```

##### Binary floating-point cannot represent every decimal number exactly.

```python
print(0.1 + 0.2)
```

#### C. Complex (complex)
A complex number contains a real part and an imaginary part. Python uses j for the imaginary unit.

```python
z = 3 + 4j

print(z)
print(type(z))
print(z.real)
print(z.imag)
```

```python
a = 2 + 3j
b = 1 + 4j

print(a + b) 
print(a * b)  
```

##### Complex numbers are useful in specialized fields such as signal processing, electrical engineering and scientific computing. You won't need them often when starting general data analysis.

```python
name = "Anurag"
course = 'CSIT'
message = """I am learning Python
for Data Science."""

print(name)
print(course)
print(message)
```

#### 6. List (list)
A list stores multiple items in a single variable. Lists are ordered, mutable, and allow duplicate values.

```python
marks = [75, 82, 90, 65, 82]

print(marks)
print(type(marks))
```

```python
# A list can contain different data types.
student = ["Anurag", 20, 3.5, True]

print(student)
```

```python
marks = [75, 82, 90, 65]
# Accessing list elements
print(marks[0])   # 75
print(marks[2])   # 90
print(marks[-1])  # 65
print(marks[1:3]) # [82, 90]
```

```python
marks = [75, 82, 90]
#List are mutuable.
marks[0] = 95

print(marks)
```

```python
# Some imp list methods.

marks = [75, 82, 90]

marks.append(88)       # Add one item
print(marks)

marks.insert(1, 70)    # Insert at index 1
print(marks)

marks.remove(90)       # Remove the first matching value
print(marks)

last_mark = marks.pop()  # Remove and return last item
print(last_mark)

marks.sort()           # Sort in ascending order
print(marks)

print(len(marks))      # Number of elements
```

```python
marks = [75, 82, 90, 65, 88]

total = sum(marks)
average = total / len(marks)

print("Total marks:", total)
print("Average marks:", average)
```

#### 7. Tuple (tuple)
A tuple is similar to a list, but it is immutable. Once created, its elements cannot be reassigned or added or removed through ordinary tuple operations.

```python
coordinates = (10, 20)
student_info = ("Anurag", 20, 3.5)

print(coordinates)
print(type(coordinates))
```

```python
coordinates = (10, 20, 30)
# Accessing tuple elements
print(coordinates[0])  # 10
print(coordinates[-1]) # 30
```

#### Why use a tuple instead of a list?
Use a tuple when the collection represents a fixed group of values that should not be changed.

```python
screen_size = (1920, 1080)

width = screen_size[0]
height = screen_size[1]

print("Width:", width)
print("Height:", height)

#Trying to modify an element raises an error
```

#### 8. Dictionary (dict)
A dictionary stores data as key-value pairs. It is especially useful when each value has a meaningful label.

Dictionaries are mutable, and their keys must be hashable. Keys are unique; assigning a value to an existing key replaces its previous value. Modern Python dictionaries preserve insertion order.

```python
student = {
    "name": "Anurag",
    "age": 20,
    "course": "CSIT",
    "gpa": 3.5
}

print(student)
print(type(student))
```

```python
# Adding and updating values

student["semester"] = 4     # Add a new key
student["gpa"] = 3.7        # Update an existing key

print(student)
```

```python
#Looping through a dictionary

for key, value in student.items():
    print(key, ":", value)
```

#### 9. Set (set)
A set is a mutable collection of unique elements. Sets do not guarantee a useful positional order for indexing, so you should not rely on their iteration order.

Sets are particularly useful for removing duplicates and performing mathematical set operations.

```python
numbers = {10, 20, 30, 20, 10}

print(numbers)
print(type(numbers))
```

```python
#Adding and removing elements

numbers = {10, 20, 30}

numbers.add(40)
print(numbers)

numbers.discard(20)
print(numbers)
```

```python
#Removing duplicates from a list

cities = ["Kathmandu", "Pokhara", "Kathmandu", "Biratnagar"]

unique_cities = set(cities)

print(unique_cities)
```
