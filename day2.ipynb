#### If Statement

```python
age = int(input('Enter your age: '))
if (age > 18):
    print('You are eligible to vote')
```

```python
name = input('Enter your name: ')
age = int(input('Enter your age: '))
if (age > 18):
    print(name, 'You are eligible to vote')
```

#### If - Else Statement

```python
age = int(input('Enter your age: '))
if (age > 18):
    print('You are eligible to vote')
else:
    print('You are not eligible to vote.')
```

#### If - Elif - Else Statement

```python
marks = int(input('Enter your marks: '))
if (marks >=0 and marks < 40):
    print('Fail!')
elif (marks >=40 and marks <=70):
    print('Pass!')
elif (marks > 70 and marks < 90):
    print('Pass, Good Marks!')
elif (marks >=90 and marks <=100):
    print('TOPPERRRR!!!!')
else:
    print('Invalid Marks!!!')
```

*Exercise 1: To check even/odd*

```python
number = int(input('Enter a positive number: '))
if(number >=0):
    if (number % 2 == 0 ):
        print(number, 'is an even number')
    else:
        print(number, 'is an odd number')
else:
    print('Enter a valid or positive integer!!')
```

*Exercise 2: Check Positive, Negative, or Zero*

```python
number = int(input('Enter a number: '))
if (number >0 ):
    print(number, 'is a positive integer')
elif(number == 0):
    print(number, 'is neither positive nor negative integer')
else:
    print(number, 'is a negative integer')
```
