#### Conditions: If/elif/else
Conditional statement allow your program to make descisions - run different parts of code based on certain conditions.

```python
age = 18
print(age>=18)
```

#### If Statement

```python
age = int(input('Enter your age: '))
if (age > 18):
    print('You are eligible to vote')
```

```python
name = input('Enter your name: ')
age = int(input('Enter your age: '))
if (age > 18):
    print(name, 'You are eligible to vote')
```

#### If - Else Statement

```python
age = int(input('Enter your age: '))
if (age > 18):
    print('You are eligible to vote')
else:
    print('You are not eligible to vote.')
```

#### If - Elif - Else Statement

```python
marks = int(input('Enter your marks: '))
if (marks >=0 and marks < 40):
    print('Fail!')
elif (marks >=40 and marks <=70):
    print('Pass!')
elif (marks > 70 and marks < 90):
    print('Pass, Good Marks!')
elif (marks >=90 and marks <=100):
    print('TOPPERRRR!!!!')
else:
    print('Invalid Marks!!!')
```

*Exercise 1: To check even/odd*

```python
number = int(input('Enter a positive number: '))
if(number >=0):
    if (number % 2 == 0 ):
        print(number, 'is an even number')
    else:
        print(number, 'is an odd number')
else:
    print('Enter a valid or positive integer!!')
```

*Exercise 2: Check Positive, Negative, or Zero*

```python
number = int(input('Enter a number: '))
if (number >0 ):
    print(number, 'is a positive integer')
elif(number == 0):
    print(number, 'is neither positive nor negative integer')
else:
    print(number, 'is a negative integer')
```

#### Loops: for/while loops
A *for* loop is used to repeat a block of code multiple times.

```python
for i in range(5):
    print("Hello")
```

```python
for i in range(3):
    print(i)
```

```python
for i in range(1, 6):
    print(i)
```

```python
for i in range(2, 11, 2):
    print(i)
```

```python
for i in range(1, 6):
    print(i * i)
```

```python
for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```

*Exercise 1: Print numbers from 1 to 10 where even numbers are multiplied by 2*

```python
for i in range (1, 10):
    if(i%2==0):
        print(i, '=', i*2)
    else:
        print(i)
```

*Exercise 2: Print the multiplication table of n number.*

```python
num = int(input('Enter a number for multiplication: '))

for i in range (1,11):
    print(num, 'x', i, '=', num*i)
```

*Exercise 3: Find the sum of n numbers.*

```python
num1 = int(input('Enter range of first number: '))
num2 = int(input('Enter range of second number: '))
total = 0
for i in range (num1, num2+1):
    total +=i
print('Sum = ', total)
```
