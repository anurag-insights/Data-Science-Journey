# Day 4: Functions and error handling

#### 1. Functions in Python
A function is a reusable block of code that performs a specific task.

```python
def greet():
    print("Hello, welcome to Python!")
def name():
    print("Anurag Sharma")

greet()
name()
```

##### Functions with parameters
A parameter is a variable that receives a value when the function is called.

```python
def greet(name):
    print("Hello", name)

greet("Anurag")
greet("Ram")
```

##### Functions with multiple parameters

```python
def add(a, b):
    print(a + b)

add(10, 20)
add(5, 15)
```

##### The return statement

```python
def add(a, b):
    print(a + b)

result = add(10, 20)
print(result)
```

```python
def calculate_average(a, b, c):
    average = (a + b + c) / 3
    return average

result = calculate_average(80, 90, 70)

print(result)
```

##### Default parameters
A function can have a default value that is used when you don't provide an argument.

```python
def greet(name="Student"):
    print("Hello", name)

greet()
greet("Anurag")
```

##### Local variables inside functions
Variables created inside a function are generally local to that function.

```python
def calculate():
    result = 10 + 20
    print(result)

calculate()
```

#### 2. Error handling in Python
An error occurs when something goes wrong while our program runs or when Python cannot execute our code as written.

```python
number = int("hello")
# This raises a ValueError because "hello" cannot be converted into an integer.
```

##### Using try and except
The try block contains code that might raise an exception. The except block handles the exception if it occurs.

```python
try:
    number = int(input("Enter a number: "))
    print(number)

except ValueError:
    print("Please enter a valid integer.")
```

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    result = a / b
    print(result)

except ZeroDivisionError:
    print("You cannot divide by zero.")

except ValueError:
    print("Please enter valid integers.")
```

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

else:
    print("Your number is", number)
```

```python
try:
    number = int("abc")

except ValueError:
    print("Conversion failed.")

finally:
    print("Program finished.")
```

```python
def check_marks(marks):
    if marks < 0:
        raise ValueError("Marks cannot be negative.")

    return marks

print(check_marks(80))
```

```python
try:
    print(check_marks(-10))

except ValueError as e:
    print(e)
```

```python
def divide(a, b):
    if b == 0:
        raise ValueError("The divisor cannot be zero.")

    return a / b


try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    result = divide(num1, num2)
    print("Result:", result)

except ValueError as e:
    print("Error:", e)
```
