#### Indexing in python
Indexing means accessing a specific element from a list using its position.

```python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits[0])  
print(fruits[1])  
print(fruits[2])  
print(fruits[3])  
```

```python
# negative indexing

print(fruits[-1])  
print(fruits[-2])  
```

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[2])
```

#### 2. Slicing in Python
Slicing means extracting a portion of a list instead of just one item.

```python
numbers = [10,20,30,40,50]
print(numbers[0:2])
print(numbers[2:6])
print(numbers[:])
print(numbers[-4:-2])

print(numbers[::2])
print(numbers[:2:-1])
```

```python
# SYNTAX: list_name[start : stop : step]
```

#### 3. List methods in Python
A list method is a built-in operation you can use to work with a list—for example, adding an item, deleting an item, sorting values, or counting occurrences.

```python
numbers = [30, 10, 20, 10]
print(numbers)
```

#### Important list methods

##### 1. append() — Add one item at the end

```python
fruits = ["apple", "banana"]

fruits.append("mango")
fruits.append('pineapple')
fruits.append(2) #integers pani append(add) hunxa.

print(fruits)
```

##### 2. extend() — Add multiple items

```python
fruits = ["apple", "banana"]

fruits.extend(["mango", "orange", 23, "pineapple"])

print(fruits)
```

##### 3. insert() — Add an item at a specific index

```python
fruits = ["apple", "orange"]

fruits.insert(1, "banana") # '1' vaneko index.
fruits.insert(2, 12)
print(fruits)
```

##### 4. remove() — Remove an item by value

```python
numbers = [10, 20, 30]

numbers.remove(20)

print(numbers)
```

##### 5. pop() — Remove and return an item

```python
numbers = [10, 20, 30]

result = numbers.pop()

print(result)  

print(numbers) 
```

##### 6. sort() — Sort items in ascending order

```python
numbers = [30, 10, 20]

numbers.sort()

print(numbers)

# sorting numbers in descending order.
numbers.sort(reverse=True)
```

##### 7. reverse() — Reverse the order

```python
numbers = [10, 20, 30]

numbers.reverse()

print(numbers)
```

##### 8. index() — Find an item's position

```python
fruits = ["apple", "banana", "mango"]

print(fruits.index("banana"))
print(fruits.index("mango"))
```

##### 9. count() — Count occurrences

```python
numbers = [10, 20, 10, 30, 10]

print(numbers.count(10))

numbers.append(10)

print(numbers.count(10))
```

##### 10. clear() — Remove all items

```python
numbers = [10, 20, 30]

numbers.clear()

print(numbers)

#clears everything in the list.
```

#### Dictionary keys and values
A dictionary in Python stores information as key-value pairs.

```python
student = {
    "name": "Anurag",
    "age": 20,
    "marks": 85
}
print(student)
```

```python
print(student["name"]) 
print(student["age"])   
print(student["marks"])
```

```python
print(student.get("name"))   
print(student.get("city"))      
print(student.get("city", "Unknown"))
print(student.get("age"))  
```

##### 1. keys() — Get all keys

```python
print(student.keys())
```

##### 2. values() — Get all values

```python
print(student.values())
```

##### 3. items() — Get keys and values together

```python
print(student.items())
```

##### Adding, updating, and deleting values

```python
student["city"] = "Biratnagar" 
student["marks"] = 90        
del student["age"]            

print(student)
```

#### Iteration in Python
Iteration means going through items one by one, usually using a for loop.

##### A. Iterating through a list

```python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit)
```

##### B. Iterating through dictionary keys

```python
student = {
    "name": "Anurag",
    "age": 20,
    "marks": 85
}

for key in student:
    print(key)

#-----OR-----
print("------------------")
for key in student.keys():
    print(key)
```

##### C. Iterating through dictionary values

```python
for value in student.values():
    print(value)
```

##### D. Iterating through keys and values together

```python
marks = {
    "Math": 80,
    "Python": 90,
    "SQL": 85
}

for subject, mark in marks.items():
    print(subject, ":", mark)
```
