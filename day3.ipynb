#### Indexing in python
Indexing means accessing a specific element from a list using its position.

```python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits[0])  
print(fruits[1])  
print(fruits[2])  
print(fruits[3])  
```

```python
# negative indexing

print(fruits[-1])  
print(fruits[-2])  
```

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[2])
```

#### 2. Slicing in Python
Slicing means extracting a portion of a list instead of just one item.

```python
numbers = [10,20,30,40,50]
print(numbers[0:2])
print(numbers[2:6])
print(numbers[:])
print(numbers[-4:-2])

print(numbers[::2])
print(numbers[:2:-1])
```
