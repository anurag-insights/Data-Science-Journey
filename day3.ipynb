#### Indexing in python
Indexing means accessing a specific element from a list using its position.

```python
fruits = ["apple", "banana", "mango", "orange"]

print(fruits[0])  
print(fruits[1])  
print(fruits[2])  
print(fruits[3])  
```

```python
# negative indexing

print(fruits[-1])  
print(fruits[-2])  
```

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[2])
```

#### 2. Slicing in Python
Slicing means extracting a portion of a list instead of just one item.

```python
numbers = [10,20,30,40,50]
print(numbers[0:2])
print(numbers[2:6])
print(numbers[:])
print(numbers[-4:-2])

print(numbers[::2])
print(numbers[:2:-1])
```

```python
# SYNTAX: list_name[start : stop : step]
```

#### 3. List methods in Python
A list method is a built-in operation you can use to work with a list—for example, adding an item, deleting an item, sorting values, or counting occurrences.

```python
numbers = [30, 10, 20, 10]
print(numbers)
```

#### Important list methods

##### 1. append() — Add one item at the end

```python
fruits = ["apple", "banana"]

fruits.append("mango")
fruits.append('pineapple')
fruits.append(2) #integers pani append(add) hunxa.

print(fruits)
```

##### 2. extend() — Add multiple items

```python
fruits = ["apple", "banana"]

fruits.extend(["mango", "orange", 23, "pineapple"])

print(fruits)
```

##### 3. insert() — Add an item at a specific index

```python
fruits = ["apple", "orange"]

fruits.insert(1, "banana") # '1' vaneko index.
fruits.insert(2, 12)
print(fruits)
```

##### 4. remove() — Remove an item by value

```python
numbers = [10, 20, 30]

numbers.remove(20)

print(numbers)
```

##### 5. pop() — Remove and return an item

```python
numbers = [10, 20, 30]

result = numbers.pop()

print(result)  

print(numbers) 
```

##### 6. sort() — Sort items in ascending order

```python
numbers = [30, 10, 20]

numbers.sort()

print(numbers)

# sorting numbers in descending order.
numbers.sort(reverse=True)
```

##### 7. reverse() — Reverse the order

```python
numbers = [10, 20, 30]

numbers.reverse()

print(numbers)
```

##### 8. index() — Find an item's position

```python
fruits = ["apple", "banana", "mango"]

print(fruits.index("banana"))
print(fruits.index("mango"))
```

##### 9. count() — Count occurrences

```python
numbers = [10, 20, 10, 30, 10]

print(numbers.count(10))

numbers.append(10)

print(numbers.count(10))
```

##### 10. clear() — Remove all items

```python
numbers = [10, 20, 30]

numbers.clear()

print(numbers)

#clears everything in the list.
```
