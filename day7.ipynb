#### One Hot Encoding and Dummy Variables.

```python
dataset = pd.read_excel(r'C:\Users\main\Desktop\data science journey\data\data_cleaning.xlsx')
```

```python
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
#filling gender missing values
dataset['Gender'] = dataset['Gender'].fillna(dataset['Gender'].mode()[0])
```

```python
dataset.isnull().sum()
```

```python
#filling married missing values
dataset['Married'] = dataset['Married'].fillna(dataset['Married'].mode()[0])
```

```python
dataset.isnull().sum()
```

#### Performing One Hot Encoding using "get dummies" (pandas)

```python
en_data = dataset[['Gender', 'Married']]
en_data
```

#### First we have to clean the incorrect and inconsistent datas in those columns.

```python
# Standardize the text to lowercase
en_data['Gender'] = en_data['Gender'].str.lower()
en_data['Married'] = en_data['Married'].str.lower()

# Fix the typos and variations 
en_data['Gender'] = en_data['Gender'].replace({'mle': 'male'})

# Assuming 'maybe' and 'yess' to just count as 'yes'
en_data['Married'] = en_data['Married'].replace({'yess': 'yes', 'maybe': 'yes'})

#  Capitalize the first letter 
en_data['Gender'] = en_data['Gender'].str.capitalize()
en_data['Married'] = en_data['Married'].str.capitalize()

# 4. Generate the dummies
en_dummies = pd.get_dummies(en_data, columns=['Gender', 'Married'])
```

```python
pd.get_dummies(en_data)
```

```python
pd.get_dummies(en_data).info()
```

#### Now using 'One Hot Encoding' with scikit learn library

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
ohe.fit_transform(en_data).toarray()
```

```python
# the above matrix is in array format but we need to convert this in dataframe.
ar = ohe.fit_transform(en_data).toarray()
pd.DataFrame(ar, columns = ['Gender_Female', 'Gender_Male', 'Married_No', 'Married_Yes'])
```
